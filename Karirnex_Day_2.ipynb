```python
bins = [0, 500000, 2000000, float ('inf')]
labels = ['Low', 'Mid', 'High']

df['revenue_tier'] = pd.cut(df['total_sales'], bins=bins, labels=labels)
```

```python
import pandas as pd

result = df.groupby(['Kota', 'category'])[
    ['total_sales', 'quantity']
].mean()

print(result)
```

```python
sales_category = df.groupby('category')['total_sales'].sum().reset_index()

sales_category = sales_category.sort_values(
    by='total_sales',
    ascending=False
)

print(sales_category)
```

```python
city_quantity = df.groupby('Kota')['quantity'].sum().reset_index()

city_quantity = city_quantity.sort_values(
    by='quantity',
    ascending=False
)

print(city_quantity)
```

```python
avg_sales = df.groupby('category')['total_sales'].mean().reset_index()

avg_sales = avg_sales.sort_values(
    by='total_sales',
    ascending=False
)

print(avg_sales)
```
